# Phase 6 - Robust Model Evaluation and Error Analysis

This notebook evaluates the selected 20-feature Random Forest on the fixed Phase 3 test set, estimates performance stability with 5-fold stratified cross-validation, compares it with the all-31-feature Random Forest, and analyzes test-set errors.

The current benchmark supports standard classification and cross-validation evaluation, but it does not support a rigorous unseen-domain evaluation because URL/domain identifiers are absent.

## Load Reproducible Data

```python
from pathlib import Path
import json

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, confusion_matrix, f1_score, precision_score,
    recall_score, roc_auc_score, roc_curve, ConfusionMatrixDisplay,
)
from sklearn.model_selection import StratifiedKFold

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
PROCESSED_DIR = PROJECT_ROOT / 'dataset' / 'processed'
RESULTS_DIR = PROJECT_ROOT / 'results'
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

X_train = pd.read_csv(PROCESSED_DIR / 'X_train_tree.csv')
X_test = pd.read_csv(PROCESSED_DIR / 'X_test_tree.csv')
y_train = pd.read_csv(PROCESSED_DIR / 'y_train.csv')['class']
y_test = pd.read_csv(PROCESSED_DIR / 'y_test.csv')['class']
selected_metadata = json.loads((RESULTS_DIR / 'selected_features.json').read_text(encoding='utf-8'))
selected_features = selected_metadata['features']
PHISHING_LABEL = -1
LEGITIMATE_LABEL = 1
FEATURE_COUNT = 20
RANDOM_STATE = 42
N_ESTIMATORS = 200
X_test_original = X_test.copy(deep=True)

print('Loaded exact Phase 3 train/test artifacts.')
print('Selected feature method:', selected_metadata['method'])
print('Selected feature count:', len(selected_features))
```

## Verify Selected Features

```python
assert X_train.shape == (7737, 31)
assert X_test.shape == (3317, 31)
assert y_train.shape == (7737,)
assert y_test.shape == (3317,)
assert len(selected_features) == FEATURE_COUNT
assert len(set(selected_features)) == FEATURE_COUNT
assert set(selected_features).issubset(set(X_train.columns))
assert set(selected_features).issubset(set(X_test.columns))
assert set(pd.concat([y_train, y_test]).unique()) == {PHISHING_LABEL, LEGITIMATE_LABEL}
assert selected_metadata['model'] == 'Random Forest'
print('Verified labels: -1 = Phishing; 1 = Legitimate')
print('X_train shape:', X_train.shape)
print('X_test shape:', X_test.shape)
print('Selected features:', selected_features)
print('The fixed test set will not be used for feature selection or CV fitting.')
```

## Reproduce Selected-Model Test Results

```python
def evaluate_model(model, X_eval, y_eval):
    predictions = model.predict(X_eval)
    classes = list(model.classes_)
    scores = model.predict_proba(X_eval)[:, classes.index(PHISHING_LABEL)]
    matrix = confusion_matrix(y_eval, predictions, labels=[LEGITIMATE_LABEL, PHISHING_LABEL])
    tn, fp, fn, tp = matrix.ravel()
    binary_target = (y_eval == PHISHING_LABEL).astype(int)
    return {
        'Accuracy': accuracy_score(y_eval, predictions),
        'Precision': precision_score(y_eval, predictions, pos_label=PHISHING_LABEL, zero_division=0),
        'Recall': recall_score(y_eval, predictions, pos_label=PHISHING_LABEL, zero_division=0),
        'F1': f1_score(y_eval, predictions, pos_label=PHISHING_LABEL, zero_division=0),
        'Specificity': tn / (tn + fp) if tn + fp else 0.0,
        'ROC_AUC': roc_auc_score(binary_target, scores),
        'FPR': fp / (fp + tn) if fp + tn else 0.0,
        'FNR': fn / (fn + tp) if fn + tp else 0.0,
        'confusion_matrix': matrix,
        'predictions': predictions,
        'scores': scores,
    }

selected_model = RandomForestClassifier(n_estimators=N_ESTIMATORS, random_state=RANDOM_STATE, n_jobs=-1)
selected_model.fit(X_train[selected_features], y_train)
selected_test = evaluate_model(selected_model, X_test[selected_features], y_test)
phase5_summary = json.loads((RESULTS_DIR / 'feature_selection_summary.json').read_text(encoding='utf-8'))
for metric in ['Accuracy', 'F1', 'ROC_AUC', 'Recall']:
    assert abs(selected_test[metric] - phase5_summary[f'selected_{metric.lower()}']) < 1e-9, metric

print('Selected-model test metrics:')
display(pd.DataFrame([{metric: selected_test[metric] for metric in ['Accuracy', 'Precision', 'Recall', 'F1', 'Specificity', 'ROC_AUC', 'FPR', 'FNR']}]))
print('Performance is consistent with the Phase 5 selected-model result.')
```

## 5-Fold Stratified Cross-Validation

The selected 20-feature CV is a complete feature-selection pipeline: in each fold, Random Forest importance is fitted on that fold's training partition, the top 20 features are selected, and a new Random Forest is fitted before evaluating the held-out fold. This prevents validation-fold information from influencing feature selection.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
cv_metric_names = ['Accuracy', 'Precision', 'Recall', 'F1', 'ROC_AUC']

def run_fold_safe_cv(select_features_per_fold):
    records = []
    fold_feature_sets = []
    for fold_number, (train_indices, validation_indices) in enumerate(cv.split(X_train, y_train), start=1):
        X_fold_train = X_train.iloc[train_indices]
        X_fold_validation = X_train.iloc[validation_indices]
        y_fold_train = y_train.iloc[train_indices]
        y_fold_validation = y_train.iloc[validation_indices]
        if select_features_per_fold:
            ranking_model = RandomForestClassifier(n_estimators=N_ESTIMATORS, random_state=RANDOM_STATE, n_jobs=-1)
            ranking_model.fit(X_fold_train, y_fold_train)
            fold_features = list(pd.Series(ranking_model.feature_importances_, index=X_train.columns).sort_values(ascending=False).head(FEATURE_COUNT).index)
        else:
            fold_features = list(X_train.columns)
        fold_model = RandomForestClassifier(n_estimators=N_ESTIMATORS, random_state=RANDOM_STATE, n_jobs=-1)
        fold_model.fit(X_fold_train[fold_features], y_fold_train)
        fold_result = evaluate_model(fold_model, X_fold_validation[fold_features], y_fold_validation)
        records.append({
            'Fold': fold_number,
            **{metric: fold_result[metric] for metric in cv_metric_names},
        })
        fold_feature_sets.append(fold_features)
    return pd.DataFrame(records), fold_feature_sets

selected_cv_results, fold_feature_sets = run_fold_safe_cv(select_features_per_fold=True)
selected_cv_results.to_csv(RESULTS_DIR / 'cross_validation_results.csv', index=False)
cv_summary_rows = []
for metric in cv_metric_names:
    cv_summary_rows.append({
        'Metric': metric,
        'Mean': selected_cv_results[metric].mean(),
        'Standard_Deviation': selected_cv_results[metric].std(ddof=1),
        'Minimum': selected_cv_results[metric].min(),
        'Maximum': selected_cv_results[metric].max(),
    })
selected_cv_summary = pd.DataFrame(cv_summary_rows)
selected_cv_summary.to_csv(RESULTS_DIR / 'cross_validation_summary.csv', index=False)
print('Selected-feature fold-safe CV results:')
display(selected_cv_results)
display(selected_cv_summary)
print('Each fold selected features from its training partition only.')
```

## Baseline vs Selected Feature CV

```python
baseline_cv_results, _ = run_fold_safe_cv(select_features_per_fold=False)

def aggregate_cv_results(results, model_name, feature_count):
    return {
        'Model': model_name,
        'Feature_Count': feature_count,
        'Mean_Accuracy': results['Accuracy'].mean(),
        'Std_Accuracy': results['Accuracy'].std(ddof=1),
        'Mean_Precision': results['Precision'].mean(),
        'Mean_Recall': results['Recall'].mean(),
        'Mean_F1': results['F1'].mean(),
        'Std_F1': results['F1'].std(ddof=1),
        'Mean_ROC_AUC': results['ROC_AUC'].mean(),
        'Std_ROC_AUC': results['ROC_AUC'].std(ddof=1),
    }

cv_comparison = pd.DataFrame([
    aggregate_cv_results(baseline_cv_results, 'Random Forest - All Features', 31),
    aggregate_cv_results(selected_cv_results, 'Random Forest - Selected Top-K', FEATURE_COUNT),
])
cv_comparison.to_csv(RESULTS_DIR / 'cv_baseline_vs_selected.csv', index=False)
display(cv_comparison)
assert list(cv_comparison['Feature_Count']) == [31, 20]
```

## Error Analysis

```python
test_predictions = pd.DataFrame({
    'Actual': y_test.to_numpy(),
    'Predicted': selected_test['predictions'],
})
test_predictions['Error_Type'] = np.select(
    [
        (test_predictions['Actual'] == PHISHING_LABEL) & (test_predictions['Predicted'] == PHISHING_LABEL),
        (test_predictions['Actual'] == LEGITIMATE_LABEL) & (test_predictions['Predicted'] == LEGITIMATE_LABEL),
        (test_predictions['Actual'] == LEGITIMATE_LABEL) & (test_predictions['Predicted'] == PHISHING_LABEL),
        (test_predictions['Actual'] == PHISHING_LABEL) & (test_predictions['Predicted'] == LEGITIMATE_LABEL),
    ],
    ['TP', 'TN', 'FP', 'FN'],
    default='Unknown',
)
assert set(test_predictions['Error_Type']) == {'TP', 'TN', 'FP', 'FN'}
test_predictions.to_csv(RESULTS_DIR / 'test_predictions.csv', index=False)
tn, fp, fn, tp = selected_test['confusion_matrix'].ravel()
error_summary = {
    'True_Positives': int(tp),
    'True_Negatives': int(tn),
    'False_Positives': int(fp),
    'False_Negatives': int(fn),
    'False_Positive_Rate': selected_test['FPR'],
    'False_Negative_Rate': selected_test['FNR'],
    'Phishing_Recall': selected_test['Recall'],
    'Specificity': selected_test['Specificity'],
}
print('Error counts:')
display(test_predictions['Error_Type'].value_counts().rename('count').to_frame())
print('Error metrics:', error_summary)
assert X_test.equals(X_test_original), 'The fixed test set was modified.'
```

## Confusion Matrix

```python
matrix_display = ConfusionMatrixDisplay(
    confusion_matrix=selected_test['confusion_matrix'],
    display_labels=['Legitimate (1)', 'Phishing (-1)'],
)
matrix_display.plot()
plt.title('Selected 20-Feature Random Forest Confusion Matrix')
plt.xlabel('Predicted label')
plt.ylabel('Actual label')
plt.tight_layout()
plt.savefig(RESULTS_DIR / 'final_confusion_matrix.png', dpi=150)
plt.show()
plt.close()
```

## ROC Curve

```python
binary_test_target = (y_test == PHISHING_LABEL).astype(int)
false_positive_rates, true_positive_rates, _ = roc_curve(binary_test_target, selected_test['scores'], pos_label=1)
plt.figure(figsize=(8, 6))
plt.plot(false_positive_rates, true_positive_rates, label=f"Selected RF (AUC={selected_test['ROC_AUC']:.4f})")
plt.plot([0, 1], [0, 1], linestyle='--', label='Chance')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Selected 20-Feature Random Forest ROC Curve')
plt.legend()
plt.tight_layout()
plt.savefig(RESULTS_DIR / 'final_roc_curve.png', dpi=150)
plt.show()
plt.close()
```

## Performance Stability

```python
for metric, filename, title in [
    ('F1', 'cv_f1_distribution.png', 'Selected-Feature CV F1 Distribution'),
    ('ROC_AUC', 'cv_roc_auc_distribution.png', 'Selected-Feature CV ROC-AUC Distribution'),
]:
    plt.figure(figsize=(7, 5))
    plt.boxplot(selected_cv_results[metric], tick_labels=['Selected RF'])
    plt.ylabel(metric)
    plt.title(title)
    plt.tight_layout()
    plt.savefig(RESULTS_DIR / filename, dpi=150)
    plt.show()
    plt.close()

plt.figure(figsize=(8, 5))
plt.bar(cv_comparison['Model'], cv_comparison['Mean_F1'])
plt.ylabel('Mean CV F1-score')
plt.title('Baseline vs Selected Feature Performance')
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(RESULTS_DIR / 'baseline_vs_selected_cv.png', dpi=150)
plt.show()
plt.close()
```

## Final Model Selection

```python
baseline_cv_row = cv_comparison[cv_comparison['Feature_Count'] == 31].iloc[0]
selected_cv_row = cv_comparison[cv_comparison['Feature_Count'] == 20].iloc[0]
reduction_stable = bool(
    selected_cv_row['Mean_F1'] >= baseline_cv_row['Mean_F1'] - 0.01
    and selected_cv_row['Mean_ROC_AUC'] >= baseline_cv_row['Mean_ROC_AUC'] - 0.01
    and selected_cv_row['Mean_Recall'] >= baseline_cv_row['Mean_Recall'] - 0.01
    and selected_cv_row['Std_F1'] <= baseline_cv_row['Std_F1'] + 0.005
)
final_model_decision = 'Random Forest with selected 20 features' if reduction_stable else 'Random Forest with all 31 features'
print('31-feature Random Forest CV:')
display(baseline_cv_row.to_frame().T)
print('20-feature Random Forest CV:')
display(selected_cv_row.to_frame().T)
print('Feature reduction remained stable:', reduction_stable)
print('Final model decision:', final_model_decision)
print('Decision considers F1, phishing recall, specificity, ROC-AUC, stability, and reduction, not accuracy alone.')
```

## Zero-Day Evaluation Limitation

The current benchmark supports standard classification and cross-validation evaluation, but it does not support a rigorous unseen-domain evaluation because URL/domain identifiers are absent. No random grouping or synthetic domain ID is used. A future dataset containing raw URLs or domain identifiers would enable domain-level train/test separation, unseen-domain testing, temporal evaluation, and stronger zero-day generalization analysis.
