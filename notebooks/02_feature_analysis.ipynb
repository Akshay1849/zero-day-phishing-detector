# 02 - Feature Analysis

This notebook analyzes the actual feature names in the dataset, groups them conceptually, reviews correlations, and computes preliminary Random Forest importance. Correlation is reported for review; features are not removed automatically.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))
from src.feature_analysis import (
    classify_columns, feature_information, high_correlation_pairs,
    infer_target_column, random_forest_importance,
)

df = pd.read_csv(PROJECT_ROOT / 'dataset' / 'raw' / 'phishing.csv')
target_column = infer_target_column(df)
results_dir = PROJECT_ROOT / 'results'
feature_table = feature_information(df, target_column)
display(feature_table)
```

```python
group_table = classify_columns(df, target_column)
display(group_table)
group_table.to_csv(results_dir / 'feature_group_classification.csv', index=False)
correlations = high_correlation_pairs(df.drop(columns=[target_column]))
display(correlations)
correlations.to_csv(results_dir / 'high_correlation_pairs.csv', index=False)
```

```python
importance = random_forest_importance(df, target_column)
display(importance)
importance.to_csv(results_dir / 'random_forest_feature_importance.csv', index=False)
print('Preliminary importance is descriptive only; it is not a final selected feature set.')
```

## Interpretation boundary

The CSV encodes URL, structural/domain, and web/security indicators as numeric values such as -1, 0, and 1. These values are treated as observed encoded features, not as raw URL text. Feature selection should be decided after baseline experiments and validation, with leakage checks and unseen-domain data considered.
