# Phase 8F-8I - Raw URL Model Training and Evaluation

This is an independent URL model. It uses only deterministic features extracted from raw URL strings and label `0 = Phishing`, `1 = Legitimate`. It does not reuse the Phase 1-7 benchmark model, features, metrics, or test set.

```python
from pathlib import Path
import json
import sys

import joblib
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit, StratifiedKFold, train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import ConfusionMatrixDisplay, roc_curve

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
from src.url_feature_extraction import URL_ONLY_FEATURE_ORDER
from src.url_modeling import make_models, train_and_evaluate, evaluate, feature_importance

RESULTS = ROOT / 'results'
MODELS = ROOT / 'models'
df = pd.read_csv(ROOT / 'dataset' / 'processed' / 'url_only_features.csv')
X = df[URL_ONLY_FEATURE_ORDER]
y = df['label']
groups = df['domain_group']
assert set(y.unique()) == {0, 1}
assert X.shape[1] == len(URL_ONLY_FEATURE_ORDER)
print('Rows:', len(df), 'URL-only features:', X.shape[1], 'domains:', groups.nunique())
print('Labels: 0 = Phishing, 1 = Legitimate')
```

## Standard and unseen-domain splits

```python
standard_train, standard_test = train_test_split(df.index, test_size=0.2, random_state=42, stratify=y)
group_splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
unseen_train, unseen_test = next(group_splitter.split(X, y, groups=groups))
unseen_train = df.index[unseen_train]
unseen_test = df.index[unseen_test]
assert set(groups.loc[unseen_train]).isdisjoint(set(groups.loc[unseen_test]))
print('Standard split:', len(standard_train), len(standard_test))
print('Unseen-domain split:', len(unseen_train), len(unseen_test))
print('Standard class proportions:', y.loc[standard_train].value_counts(normalize=True).to_dict())
print('Unseen train/test labels:', y.loc[unseen_train].value_counts().to_dict(), y.loc[unseen_test].value_counts().to_dict())
print('Domains train/test:', groups.loc[unseen_train].nunique(), groups.loc[unseen_test].nunique())
```

## Classifier comparison

```python
standard_models = make_models()
standard_results, standard_fitted = train_and_evaluate(
    standard_models, X.loc[standard_train], X.loc[standard_test], y.loc[standard_train], y.loc[standard_test]
)
unseen_models = make_models()
unseen_results, unseen_fitted = train_and_evaluate(
    unseen_models, X.loc[unseen_train], X.loc[unseen_test], y.loc[unseen_train], y.loc[unseen_test]
)
standard_results['Evaluation'] = 'Standard stratified split'
unseen_results['Evaluation'] = 'Unseen-domain split'
url_model_results = pd.concat([standard_results, unseen_results], ignore_index=True)
url_model_results.to_csv(RESULTS / 'url_model_results.csv', index=False)
display(url_model_results[['Evaluation','Model','Accuracy','Precision','Recall','F1','Specificity','ROC_AUC','FPR','FNR']])
```

## URL feature importance and feature-count analysis

```python
ranking_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
ranking_model.fit(X.loc[standard_train], y.loc[standard_train])
importance = feature_importance(ranking_model, URL_ONLY_FEATURE_ORDER)
importance.to_csv(RESULTS / 'url_feature_importance.csv', index=False)
ranked = importance['feature'].tolist()
feature_counts = [5, 10, 15, 20, len(ranked)]
selection_rows = []
for count in feature_counts:
    selected = ranked[:count]
    candidate = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
    candidate.fit(X.loc[standard_train, selected], y.loc[standard_train])
    standard_metrics = evaluate(candidate, X.loc[standard_test, selected], y.loc[standard_test])
    unseen_candidate = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
    unseen_candidate.fit(X.loc[unseen_train, selected], y.loc[unseen_train])
    unseen_metrics = evaluate(unseen_candidate, X.loc[unseen_test, selected], y.loc[unseen_test])
    selection_rows.append({'Feature_Count': count, 'Features': '|'.join(selected), 'Standard_F1': standard_metrics['F1'], 'Standard_ROC_AUC': standard_metrics['ROC_AUC'], 'Unseen_F1': unseen_metrics['F1'], 'Unseen_ROC_AUC': unseen_metrics['ROC_AUC'], 'Unseen_Recall': unseen_metrics['Recall'], 'Unseen_Specificity': unseen_metrics['Specificity']})
selection_results = pd.DataFrame(selection_rows)
selection_results.to_csv(RESULTS / 'url_feature_selection.csv', index=False)
display(selection_results)
```

## Final URL model and 5-fold standard cross-validation

The feature set is selected by unseen-domain F1, then unseen-domain ROC-AUC, then smaller feature count. This explicitly prioritizes generalization and complexity rather than random-split accuracy alone.

```python
chosen = selection_results.sort_values(['Unseen_F1','Unseen_ROC_AUC','Feature_Count'], ascending=[False,False,True]).iloc[0]
selected_features = chosen['Features'].split('|')
final_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
final_model.fit(X.loc[standard_train, selected_features], y.loc[standard_train])
final_standard = evaluate(final_model, X.loc[standard_test, selected_features], y.loc[standard_test])
final_unseen_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
final_unseen_model.fit(X.loc[unseen_train, selected_features], y.loc[unseen_train])
final_unseen = evaluate(final_unseen_model, X.loc[unseen_test, selected_features], y.loc[unseen_test])
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_rows = []
for fold, (train_idx, valid_idx) in enumerate(cv.split(X.loc[standard_train], y.loc[standard_train]), start=1):
    fold_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
    fold_model.fit(X.loc[standard_train].iloc[train_idx][selected_features], y.loc[standard_train].iloc[train_idx])
    metrics = evaluate(fold_model, X.loc[standard_train].iloc[valid_idx][selected_features], y.loc[standard_train].iloc[valid_idx])
    cv_rows.append({'Fold': fold, **{key: metrics[key] for key in ['Accuracy','Precision','Recall','F1','ROC_AUC']}})
cv_results = pd.DataFrame(cv_rows)
cv_results.to_csv(RESULTS / 'url_cross_validation_results.csv', index=False)
cv_summary = {metric: {'mean': float(cv_results[metric].mean()), 'std': float(cv_results[metric].std(ddof=1))} for metric in ['Accuracy','Precision','Recall','F1','ROC_AUC']}
summary = {'dataset_rows': int(len(df)), 'unique_domains': int(groups.nunique()), 'url_only_feature_count': int(len(URL_ONLY_FEATURE_ORDER)), 'selected_feature_count': int(len(selected_features)), 'selected_features': selected_features, 'label_mapping': {'0':'Phishing','1':'Legitimate'}, 'standard_split': {'test_rows': int(len(standard_test)), **{k: float(v) for k,v in final_standard.items() if k in ['Accuracy','Precision','Recall','F1','Specificity','ROC_AUC','FPR','FNR']}}, 'unseen_domain_split': {'test_rows': int(len(unseen_test)), 'train_domains': int(groups.loc[unseen_train].nunique()), 'test_domains': int(groups.loc[unseen_test].nunique()), **{k: float(v) for k,v in final_unseen.items() if k in ['Accuracy','Precision','Recall','F1','Specificity','ROC_AUC','FPR','FNR']}}, 'cross_validation': cv_summary, 'validation_note': 'Standard CV is stratified. Unseen-domain evaluation uses disjoint registrable-domain groups.', 'model': 'Random Forest n_estimators=100 random_state=42'}
(RESULTS / 'url_model_summary.json').write_text(json.dumps(summary, indent=2), encoding='utf-8')
(RESULTS / 'url_selected_features.json').write_text(json.dumps({'features': selected_features, 'feature_count': len(selected_features), 'selection_criterion': 'highest unseen-domain F1, then ROC-AUC, then fewer features'}, indent=2), encoding='utf-8')
joblib.dump(final_model, MODELS / 'url_final_model.pkl')
(MODELS / 'url_feature_order.json').write_text(json.dumps(selected_features, indent=2), encoding='utf-8')
print('Selected feature count:', len(selected_features))
print('Standard metrics:', final_standard)
print('Unseen-domain metrics:', final_unseen)
print('CV summary:', cv_summary)
```

## Curves and comparison figures

```python
plt.figure(figsize=(9,5))
plt.bar(url_model_results['Model'] + ' - ' + url_model_results['Evaluation'], url_model_results['F1'])
plt.xticks(rotation=75, ha='right')
plt.ylabel('F1')
plt.tight_layout()
plt.savefig(RESULTS / 'url_model_comparison.png', dpi=150)
plt.close()
plt.figure(figsize=(8,5))
for name, model in standard_fitted.items():
    scores = model.predict_proba(X.loc[standard_test])[:, list(model.classes_).index(0)] if hasattr(model, 'predict_proba') else -model.decision_function(X.loc[standard_test])
    fpr, tpr, _ = roc_curve((y.loc[standard_test] == 0).astype(int), scores)
    from sklearn.metrics import roc_auc_score
    plt.plot(fpr, tpr, label=f'{name} (AUC={roc_auc_score((y.loc[standard_test] == 0).astype(int), scores):.3f})')
plt.plot([0,1],[0,1],'--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend()
plt.tight_layout()
plt.savefig(RESULTS / 'url_roc_curve.png', dpi=150)
plt.close()
ConfusionMatrixDisplay.from_predictions(y.loc[standard_test], standard_fitted['Random Forest'].predict(X.loc[standard_test]), labels=[1,0], display_labels=['Legitimate','Phishing'])
plt.tight_layout()
plt.savefig(RESULTS / 'url_confusion_matrix.png', dpi=150)
plt.close()
print('Saved URL model figures.')
```
